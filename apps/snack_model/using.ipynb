```python
import os
import torch
import cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO
from PIL import Image

# Select device
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}")

# Load the trained YOLO model
model_path = "apps/snack_model/best.pt"
folder_path = "../downloads"  # Update this to your folder of images

# Ensure the folder exists
if not os.path.exists(folder_path):
    raise FileNotFoundError(f"Error: Folder not found at {folder_path}")

# List image files with common extensions
image_extensions = (".jpg", ".jpeg", ".png", ".bmp")
image_files = [os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.lower().endswith(image_extensions)]

if not image_files:
    raise FileNotFoundError("No image files found in the folder.")

# Load YOLO model and move to the appropriate device
model = YOLO(model_path).to(device)

# Process each image in the folder
for image_path in image_files:
    print(f"Processing: {image_path}")
    
    # Run inference on the current image with a given confidence threshold
    results = model(image_path, conf=0.2)
    
    # Display detection results
    for r in results:
        im_array_bgr = r.plot()  # Draw bounding boxes (in BGR format)
        im_array_rgb = cv2.cvtColor(im_array_bgr, cv2.COLOR_BGR2RGB)  # Convert to RGB
        im = Image.fromarray(im_array_rgb)
        
        plt.figure(figsize=(10, 10))
        plt.imshow(im)
        plt.axis("off")
        plt.title(f"YOLO Detection Results for {os.path.basename(image_path)}")
        plt.show()
```
